```python
import random
import time

class Environment:
    def __init__(self):
        self.locations = {
            'A': random.choice(['Clean', 'Dirty']),
            'B': random.choice(['Clean', 'Dirty'])
        }
        self.agent_location = random.choice(['A', 'B'])
        # Added a counter to track the total cost
        self.total_cost = 0

    def get_percept(self):
        return self.agent_location, self.locations[self.agent_location]

    def execute_action(self, action):
        print(f"Agent is at Location {self.agent_location} | Status: {self.locations[self.agent_location]} | Action taken: {action}")

        # Calculate costs based on the action taken
        if action == 'Suck':
            self.locations[self.agent_location] = 'Clean'
            self.total_cost += 2  # Cleaning takes more energy
        elif action == 'Right':
            self.agent_location = 'B'
            self.total_cost += 1  # Moving takes energy
        elif action == 'Left':
            self.agent_location = 'A'
            self.total_cost += 1  # Moving takes energy
        elif action == 'NoOp':
            pass  # Idiing costs 0 points

    def are_all_locations_clean(self):
        return self.locations['A'] == 'Clean' and self.locations['B'] == 'Clean'


class SimpleReflexVacuumAgent:
    def reflex_agent_program(self, percept):
        location, status = percept

        if status == 'Dirty':
            return 'Suck'
        elif location == 'A':
            return 'Right'
        elif location == 'B':
            return 'Left'

        return 'NoOp'


# --- Simulation Runner ---
if __name__ == "__main__":
    env = Environment()
    agent = SimpleReflexVacuumAgent()
    print("Name: H G Shreyas")
    print("Usn: 1BM24CS108")
    print("Initial Environment State:", env.locations)
    print(f"Vacuum starting at location: {env.agent_location}\n" + "-"*40)

    steps = 0
    max_steps = 10

    while steps < max_steps:
        current_percept = env.get_percept()
        chosen_action = agent.reflex_agent_program(current_percept)
        env.execute_action(chosen_action)

        if env.are_all_locations_clean() and chosen_action == 'NoOp':
            print("-"*40 + "\nAll locations are clean! Stopping agent.")
            break

        steps += 1
        time.sleep(0.5)

    # Print the final cost calculation at the end
    print(f"FINAL PERFORMANCE SCORE -> Total Cost Incurred: {env.total_cost} points")
```
