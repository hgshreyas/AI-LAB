```python

import time


class Environment:
    def __init__(self):
        # Take user input for the environment
        self.locations = {}

        for location in ['A', 'B']:
            while True:
                status = input(
                    f"Enter status of Location {location} (Clean/Dirty): "
                ).strip().capitalize()

                if status in ['Clean', 'Dirty']:
                    self.locations[location] = status
                    break
                else:
                    print("Invalid input! Enter Clean or Dirty.")

        # Take initial vacuum cleaner position
        while True:
            position = input(
                "Enter initial vacuum cleaner position (A/B): "
            ).strip().upper()

            if position in ['A', 'B']:
                self.agent_location = position
                break
            else:
                print("Invalid input! Enter A or B.")

        self.total_cost = 0

    def get_percept(self):
        return (
            self.agent_location,
            self.locations[self.agent_location]
        )

    def execute_action(self, action):
        print(
            f"Agent is at Location {self.agent_location} | "
            f"Status: {self.locations[self.agent_location]} | "
            f"Action taken: {action}"
        )

        if action == 'Suck':
            self.locations[self.agent_location] = 'Clean'
            self.total_cost += 1

        elif action == 'Right':
            self.agent_location = 'B'
            self.total_cost += 1

        elif action == 'Left':
            self.agent_location = 'A'
            self.total_cost += 1

        elif action == 'NoOp':
            pass

    def are_all_locations_clean(self):
        return (
            self.locations['A'] == 'Clean'
            and self.locations['B'] == 'Clean'
        )


class SimpleReflexVacuumAgent:
    def reflex_agent_program(self, percept):
        location, status = percept

        if status == 'Dirty':
            return 'Suck'

        elif location == 'A':
            return 'Right'

        elif location == 'B':
            return 'Left'

        return 'NoOp'


# --- Simulation Runner ---

if __name__ == "__main__":

    print("Name: H G Shreyas")
    print("USN: 1BM24CS108")
    print("-" * 40)

    env = Environment()
    agent = SimpleReflexVacuumAgent()

    print("\nInitial Environment State:", env.locations)
    print(f"Vacuum starting at location: {env.agent_location}")
    print("-" * 40)

    steps = 0
    max_steps = 10

    while steps < max_steps:

        # Stop when both locations are clean
        if env.are_all_locations_clean():
            print("All locations are clean! Stopping agent.")
            break

        current_percept = env.get_percept()

        chosen_action = agent.reflex_agent_program(
            current_percept
        )

        env.execute_action(chosen_action)

        steps += 1
        time.sleep(0.5)

    print("-" * 40)
    print("Final Environment State:", env.locations)
    print(f"Final Vacuum Position: {env.agent_location}")

    print(
        f"FINAL PERFORMANCE SCORE -> "
        f"Total Cost Incurred: {env.total_cost} points"
    )
```
