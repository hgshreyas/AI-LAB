```python
import random
import time

def print_grid(grid):
    """Displays the board. 0-8 layout."""
    print(f"\n {grid[0]} | {grid[1]} | {grid[2]} ")
    print("-----------")
    print(f" {grid[3]} | {grid[4]} | {grid[5]} ")
    print("-----------")
    print(f" {grid[6]} | {grid[7]} | {grid[8]} \n")

def check_win(grid, player):
    """Checks rows, columns, and diagonals."""
    lines = [
        (0,1,2), (3,4,5), (6,7,8), # Rows
        (0,3,6), (1,4,7), (2,5,8), # Cols
        (0,4,8), (2,4,6)           # Diagonals
    ]
    for a,b,c in lines:
        if grid[a] == grid[b] == grid[c] == player:
            return True
    return False

def get_ai_move(grid):
    """AI Logic: 1. Win? 2. Block Human? 3. Random."""
    possible_moves = [i for i, spot in enumerate(grid) if spot == ' ']

    # 1. Try to WIN immediately
    for move in possible_moves:
        grid_copy = grid[:]
        grid_copy[move] = 'O'
        if check_win(grid_copy, 'O'):
            return move

    # 2. Block YOU if you are about to win
    for move in possible_moves:
        grid_copy = grid[:]
        grid_copy[move] = 'X'
        if check_win(grid_copy, 'X'):
            return move

    # 3. Pick Random
    return random.choice(possible_moves)

def play_game():
    grid = [' ']*9
    print("Name: H G Shreyas")
    print("Usn: 1BM24CS108")
    print("Tic-Tac-Toe: You are X, AI is O.")
    print("Enter positions 0-8 to play.")
    print_grid(grid)

    while True:
        # --- HUMAN TURN (X) ---
        while True:
            try:
                user_input = int(input("Your Move (X): "))
                if 0 <= user_input <= 8 and grid[user_input] == ' ':
                    grid[user_input] = 'X'
                    break # Valid move, exit input loop
                else:
                    print("Position taken or invalid. Try again.")
            except ValueError:
                print("Please enter a number 0-8.")

        print_grid(grid)

        if check_win(grid, 'X'):
            print("🎉 YOU WIN! 🎉")
            break
        if ' ' not in grid:
            print("Draw Game!")
            break

        # --- AI TURN (O) ---
        # No input() here. It runs automatically.
        print("AI (O) is thinking...")
        time.sleep(1) # Small delay so it feels like thinking

        ai_move = get_ai_move(grid)
        grid[ai_move] = 'O' # AI places move

        print(f"AI plays at position: {ai_move}")
        print_grid(grid)

        if check_win(grid, 'O'):
            print("🤖 AI WINS! 🤖")
            break
        if ' ' not in grid:
            print("Draw Game!")
            break

if __name__ == "__main__":
    play_game()
```
